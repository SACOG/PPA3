```python
import geopandas as gpd
from pathlib import Path
congestion_update_path= Path(r"I:\Projects\Josh\Layer_update\Congestion")  # base path for this year's update
ca_nhs = gpd.read_file(congestion_update_path / "California.shp" ) #Download each year from here: https://npmrds.ritis.org/analytics/shapefiles
sacog = gpd.read_file(r"I:\Projects\Josh\Geospatial Data\GISOWNER\PlanningArea\PlanningArea.shp") #Use any Shapefile of SACOG planning area, to clip NHS to just SACOG region

# Make sure both are in the same CRS first
sacog = sacog.to_crs(ca_nhs.crs)

# spatial join with 'intersects' predicate keeps any TMC touching SACOG
sacog_nhs = gpd.sjoin(ca_nhs, sacog[["geometry"]], predicate="intersects", how="inner")
sacog_nhs = sacog_nhs.drop(columns=["index_right"])
sacog_nhs.to_file(layer_update_path / "Congestion")
```
